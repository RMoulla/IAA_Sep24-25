```python
import tiktoken


enc = tiktoken.encoding_for_model("gpt-5")

tokens = enc.encode("libellule")

print(tokens)

for t in tokens:
    print(t, repr(enc.decode([t])))
```

```python
enc = tiktoken.get_encoding("o200k_base")

print(len(enc._mergeable_ranks))
```

```python
print(enc._special_tokens)
```

```python
for k, v in sorted(enc._special_tokens.items(), key=lambda x: x[1]):
    print(v, k)
```

```python

ids = set(enc._mergeable_ranks.values()) | set(enc._special_tokens.values())

missing = [i for i in range(max(ids) + 1) if i not in ids]

print(missing)
print(len(missing))
```
